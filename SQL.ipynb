## SQL Table Creation

```sql
CREATE TABLE Employees (
    Emp_id INT PRIMARY KEY,
    Emp_Name VARCHAR(255) NOT NULL,
    Salary INT,
    Location VARCHAR(100),
    Phone_Number VARCHAR(20),
    Aadhar_Number VARCHAR(12) UNIQUE
);
```

How do you retrieve all records from a table?

To retrieve all records from a table, you use the `SELECT` statement with `*` (which means all columns) and the `FROM` keyword followed by the table name.

```sql
SELECT * FROM Employees;
```

How do you filter records using WHERE?

To filter records in SQL, you use the `WHERE` clause after the `FROM` clause. You can specify various conditions using comparison operators (`=`, `!=`, `<`, `>`, `<=`, `>=`) and logical operators (`AND`, `OR`, `NOT`).

For example, to retrieve employees with a salary greater than 50000:

```sql
SELECT Emp_id, Emp_Name, Salary
FROM Employees
WHERE Salary > 50000;
```

Or, to find employees from a specific location:

```sql
SELECT * FROM Employees
WHERE Location = 'New York';
```

How do you sort records using ORDER BY?

To sort records in SQL, you use the `ORDER BY` clause. This clause is placed after the `FROM` or `WHERE` clause.

You can specify the column(s) by which you want to sort, and the order (ascending or descending).

*   `ASC`: Ascending order (default if not specified).
*   `DESC`: Descending order.

### Example 1: Sort by Salary in ascending order

```sql
SELECT Emp_id, Emp_Name, Salary
FROM Employees
ORDER BY Salary ASC;
```

### Example 2: Sort by Salary in descending order

```sql
SELECT Emp_id, Emp_Name, Salary
FROM Employees
ORDER BY Salary DESC;
```

### Example 3: Sort by Location (ascending) and then by Salary (descending)

```sql
SELECT Emp_id, Emp_Name, Location, Salary
FROM Employees
ORDER BY Location ASC, Salary DESC;
```

How do you find the highest salary from an employee table?

To find the highest salary in the `Employees` table, you can use the `MAX()` aggregate function on the `Salary` column.

```sql
SELECT MAX(Salary) AS HighestSalary
FROM Employees;
```

This query will return a single value representing the maximum salary found in the `Salary` column of the `Employees` table. `AS HighestSalary` is used to give a meaningful name to the result column.

How do you find the second-highest salary?

There are several ways to find the second-highest salary in a table using SQL. Here are a few common methods:

### Method 1: Using `ORDER BY`, `LIMIT`, and `OFFSET` (or `TOP` for SQL Server)

This is often the most straightforward method for databases that support `LIMIT` and `OFFSET` (like MySQL, PostgreSQL, SQLite).

```sql
SELECT Salary
FROM Employees
ORDER BY Salary DESC
LIMIT 1 OFFSET 1;
```

*   `ORDER BY Salary DESC`: Sorts salaries in descending order (highest to lowest).
*   `OFFSET 1`: Skips the first row (the highest salary).
*   `LIMIT 1`: Returns only the next row (the second highest salary).

### Method 2: Using a Subquery with `MAX()` and `WHERE`

This method finds the maximum salary that is less than the overall maximum salary.

```sql
SELECT MAX(Salary)
FROM Employees
WHERE Salary < (SELECT MAX(Salary) FROM Employees);
```

*   The inner subquery `(SELECT MAX(Salary) FROM Employees)` finds the absolute highest salary.
*   The outer query then finds the maximum salary among all salaries that are strictly less than the absolute highest.

### Method 3: Using `DENSE_RANK()` (Window Function)

Window functions are powerful and provide a flexible way to handle ranking. `DENSE_RANK()` assigns a rank within a partition of a result set, with no gaps in the ranking sequence.

```sql
SELECT Salary
FROM (
    SELECT
        Salary,
        DENSE_RANK() OVER (ORDER BY Salary DESC) as rank_num
    FROM Employees
) AS ranked_salaries
WHERE rank_num = 2;
```

*   `DENSE_RANK() OVER (ORDER BY Salary DESC)` assigns a rank to each salary, where the highest salary gets rank 1, the second highest gets rank 2, and so on. If there are ties, they receive the same rank, and the next distinct salary receives the next consecutive rank.
*   The outer query then selects the salary where `rank_num` is 2.
