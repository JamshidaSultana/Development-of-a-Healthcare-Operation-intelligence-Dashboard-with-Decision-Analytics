Write Python code to check whether a given number is even or odd.

Requirements:

1. Take the number as input from the user.
2. Use the modulo operator (%).
3. Print "Even" if the number is divisible by 2.
4. Otherwise, print "Odd".
5. Keep the code simple and suitable for a beginner.
6. Explain the code briefly.


```python
# Take input from the user
number = int(input("Enter an integer: "))

# Check if the number is even or odd using the modulo operator
if number % 2 == 0:
    print("Even")
else:
    print("Odd")
```

### Code Explanation:
1.  **`number = int(input("Enter an integer: "))`**: This line prompts the user to enter an integer. The `input()` function reads the input as a string, and `int()` converts it into an integer, storing it in the `number` variable.
2.  **`if number % 2 == 0:`**: This is the core logic. The modulo operator (`%`) returns the remainder of a division. If a number is perfectly divisible by 2, the remainder will be 0. This `if` statement checks if the remainder is 0.
3.  **`print("Even")`**: If the condition in the `if` statement is true (i.e., the remainder is 0), it means the number is even, so "Even" is printed.
4.  **`else: print("Odd")`**: If the `if` condition is false (i.e., the remainder is not 0), it means the number is odd, and "Odd" is printed.

Write Python code to find the largest number from a given list of numbers.

Requirements:

1. Create a list containing at least 5 numbers.
2. Find the largest number in the list.
3. Use a simple Python approach suitable for a beginner.
4. Display the largest number as the output.
5. Explain the code briefly.

```python
# Create a list of numbers
numbers = [10, 5, 20, 15, 30, 75]

# Find the largest number using the max() function
largest_number = max(numbers)

# Display the largest number
print(f"The largest number in the list is: {largest_number}")
```

### Code Explanation:
1.  **`numbers = [10, 5, 20, 15, 30, 25]`**: This line initializes a list called `numbers` with several integer values.
2.  **`largest_number = max(numbers)`**: Python's built-in `max()` function is used here. It takes an iterable (like a list) as an argument and returns the largest item in it. The result is stored in the `largest_number` variable.
3.  **`print(f"The largest number in the list is: {largest_number}")`**: This line uses an f-string to print a descriptive message along with the value of the `largest_number`.

How do you find a second largest number in the list

```python
# To find the second largest number, we can sort the list in descending order
# and then pick the element at index 1 (the second element).

# Make a copy of the list to avoid modifying the original if it's used elsewhere
sorted_numbers = sorted(numbers, reverse=True)

# Check if there are at least two numbers in the list
if len(sorted_numbers) >= 2:
    second_largest_number = sorted_numbers[1]
    print(f"The second largest number in the list is: {second_largest_number}")
else:
    print("The list does not contain enough elements to find a second largest number.")
```

### Code Explanation:
1.  **`sorted_numbers = sorted(numbers, reverse=True)`**: This line creates a new list called `sorted_numbers` by sorting the original `numbers` list. The `reverse=True` argument ensures that the list is sorted in descending order (from largest to smallest).
2.  **`if len(sorted_numbers) >= 2:`**: This checks if the sorted list has at least two elements. We need at least two elements to be able to find a second largest number.
3.  **`second_largest_number = sorted_numbers[1]`**: If there are enough elements, this line accesses the element at index `1` of the `sorted_numbers` list. Since the list is sorted in descending order, the element at index `0` is the largest, and the element at index `1` is the second largest.
4.  **`print(f"The second largest number in the list is: {second_largest_number}")`**: This prints the found second largest number.
5.  **`else: print(...)`**: If the list has fewer than two elements, a message is printed indicating that a second largest number cannot be found.

How do you reverse a string in python
The given string is "INFOSYS INTERNSHIP"

```python
# The given string
original_string = "INFOSYS SPRINGBOARD"

# Reverse the string using slicing
reversed_string = original_string[::-1]

# Display the reversed string
print(f"Original string: {original_string}")
print(f"Reversed string: {reversed_string}")
```

### Code Explanation:
1.  **`original_string = "INFOSYS INTERNSHIP"`**: This line initializes a string variable `original_string` with the value "INFOSYS INTERNSHIP".
2.  **`reversed_string = original_string[::-1]`**: This is the most common and Pythonic way to reverse a string. It uses string slicing:
    *   `:` before the second colon means start from the beginning of the string.
    *   `:` after the first colon and before the second colon means go to the end of the string.
    *   `-1` is the step value, which tells Python to step backward one character at a time. This effectively reverses the string.
3.  **`print(f"Original string: {original_string}")`**: This line prints the original string.
4.  **`print(f"Reversed string: {reversed_string}")`**: This line prints the newly created reversed string.

How do count the number of vowels in a string

```python
# Define the string to count vowels from
input_string = "INFOSYS SPRINGBOARD".lower() # Convert to lowercase for easier counting

# Define a set of vowels for quick lookup
vowels = {'a', 'e', 'i', 'o', 'u'}

# Initialize a counter for vowels
vowel_count = 0

# Iterate through each character in the string
for char in input_string:
    if char in vowels:
        vowel_count += 1

# Display the total vowel count
print(f"The number of vowels in the string '{input_string.upper()}' is: {vowel_count}")
```

### Code Explanation:
1.  **`input_string = "INFOSYS SPRINGBOARD".lower()`**: We start by defining our `input_string`. To make vowel counting case-insensitive, we convert the entire string to lowercase using `.lower()`. This means we only need to check for lowercase vowels.
2.  **`vowels = {'a', 'e', 'i', 'o', 'u'}`**: A `set` named `vowels` is created. Using a set for vowels is highly efficient for checking if a character is present within it, especially for larger sets or frequent lookups, as it provides average O(1) (constant time) complexity.
3.  **`vowel_count = 0`**: An integer variable `vowel_count` is initialized to `0`. This variable will store the total number of vowels found.
4.  **`for char in input_string:`**: This loop iterates through each character (`char`) in our `input_string`.
5.  **`if char in vowels:`**: Inside the loop, for each character, we check if it exists in our `vowels` set. If it does, the character is a vowel.
6.  **`vowel_count += 1`**: If the character is a vowel, we increment our `vowel_count` by 1.
7.  **`print(f"...")`**: Finally, an f-string is used to print the total count of vowels found in the original string (converted back to uppercase for display clarity).

How do you remove the duplicate elements from the list
The given list is [ 1,3,2,2,4,6,7,5,5,5,8,9,9,10]

```python
# The given list with duplicate elements
original_list = [1, 3, 2, 2, 4, 6, 7, 5, 5, 5, 8, 9, 9, 10]

# Convert the list to a set to remove duplicates, then convert it back to a list
list_without_duplicates = list(set(original_list))

# Display the list without duplicates
print(f"Original list: {original_list}")
print(f"List without duplicates: {list_without_duplicates}")
```

### Code Explanation:
1.  **`original_list = [1, 3, 2, 2, 4, 6, 7, 5, 5, 5, 8, 9, 9, 10]`**: This line initializes `original_list` with the given numbers, including duplicates.
2.  **`set(original_list)`**: This is the key step. When you convert a list to a `set` in Python, the set automatically discards all duplicate elements, as sets by definition only contain unique values.
3.  **`list(...)`**: After creating a set with unique elements, we convert it back to a list using the `list()` constructor. The order of elements might not be preserved from the original list when converting via a set, but all duplicates will be gone.
4.  **`print(...)`**: These lines display both the `original_list` and the `list_without_duplicates` for comparison.

How do you count the frequency of each element in a list

```python
from collections import Counter

# The given list
my_list = [1, 3, 2, 2, 4, 6, 7, 5, 5, 5, 8, 9, 9, 10]

# Count the frequency of each element
frequency_counter = Counter(my_list)

# Display the frequencies
print(f"Original list: {my_list}")
print(f"Element frequencies: {frequency_counter}")
```

### Code Explanation:
1.  **`from collections import Counter`**: This line imports the `Counter` class from Python's `collections` module. The `Counter` is a specialized dictionary subclass for counting hashable objects.
2.  **`my_list = [1, 3, 2, 2, 4, 6, 7, 5, 5, 5, 8, 9, 9, 10]`**: This defines the list for which we want to count element frequencies.
3.  **`frequency_counter = Counter(my_list)`**: This is the core of the solution. When you pass a list (or any iterable) to the `Counter` constructor, it automatically iterates through the elements and creates a dictionary-like object where keys are the unique elements from the list and values are their respective frequencies (counts).
4.  **`print(...)`**: These lines display the original list and the resulting `frequency_counter` object, which clearly shows each element and how many times it appeared in the list.
